## Voronoi partition and its variants

`allocation_partitions` draws the canonical Voronoi assignment next to the fixed IDR partition built on
the public-activity field and the matched IDR partition built on one candidate field (1x3, territories
coloured per station). `partition_departure` gives the share of cells each variant assigns differently
from Voronoi; the G0/G1 gate outcomes and transport masses are in the `partition_summary` table. CIVD is
retired and not part of this comparison.


```python
REGION = None  # None = first configured analysis region; set a region id to switch
display(tables.partition_summary(ctx, region=REGION))
fig, path = figures.allocation_partitions(ctx, region=REGION)
display(fig)
plt.close(fig)
print(path)
fig, path = figures.partition_departure(ctx, region=REGION)
display(fig)
plt.close(fig)
print(path)
```
