# NZ 07 · 分叶哈希门（leaf hash gate）

读取既有清单，按分区回比文件并报告。首次生成只写视图目录，报告“首次生成，无比对”，不判 PASS。
封存根清单须经作者确认后放入；PASS 仅证明清单覆盖范围内逐字节一致。


```python
import os
from pathlib import Path
from sglib.core.infra.paths import find_repo_root
from sglib.generator import stage
from sglib.generator.manifest import run_gate, write_setup

REPO = find_repo_root(Path.cwd())
COUNTRY = "nz"
PROFILE = os.environ.get("SG_PROFILE", "formal")
RESULTS_ROOT = os.environ.get("SG_RESULTS_ROOT")
ctx = stage.country_context(REPO, COUNTRY, profile=PROFILE, results_root=RESULTS_ROOT)
print(f"results_root={ctx.results} read_only={ctx.read_only}")
```

```python
write_setup(ctx)
report = run_gate(ctx)
if report["status"] == "FAIL":
    raise SystemExit(1)
if report["status"] == "PASS" and (ctx.root / "audit.json").is_file():
    from IPython.display import display
    from sglib.generator.overview import tables
    display(tables.audit(ctx))
```
