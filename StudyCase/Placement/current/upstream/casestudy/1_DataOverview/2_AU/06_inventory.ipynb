# AU 06 · Inventory 哈希门（hash gate）

读取既有 inventory，仅比较 artifacts 声明路径的缺失与内容变化。
无既有清单时只生成到视图目录，报告“首次生成，无比对”，不判 PASS。


```python
from sglib.core.infra.paths import find_repo_root
from sglib.dataoverview import stage
from sglib.dataoverview.manifest import run_gate, write_setup

ROOT = find_repo_root()
COUNTRY = "au"
context = stage.country_context(ROOT, COUNTRY)
write_setup(context)
report = run_gate(context)
if report["status"] == "FAIL":
    raise SystemExit(1)
```

```python
if report["status"] == "PASS":
    import json
    import pandas as pd
    from IPython.display import display
    from sglib.dataoverview.overview import tables
    directory = context.merged["country"]["directory"]
    inventory_path = ROOT / "results/1_DataOverview" / directory / "data_inventory.json"
    inventory = json.loads(inventory_path.read_text(encoding="utf-8"))
    display(pd.Series({key: value for key, value in inventory.items()
                       if key not in {"regions", "artifacts", "handoff_artifacts"}},
                      name="value").to_frame())
    display(tables.inventory_regions(inventory))
    display(tables.inventory_evidence(inventory))
    display(tables.inventory_artifacts(inventory))
```
